```python
import matplotlib.pyplot as plt

# 在原来5.1的基础上修改函数，增加计数功能
def Find_expression(target, start=0, total=0, expr=""):
    a = "123456789"

    # 当所有数字都使用完毕时，判断是否满足目标值
    if start == len(a):
        if total == target:
            return 1  # 找到一个符合条件的表达式
        return 0  # 不符合条件

    count = 0  # 记录满足条件的表达式数量

    for end in range(start + 1, len(a) + 1):
        part = a[start:end]
        num = int(part)

        if start == 0:
            count += Find_expression(target, end, num, part)
        else:
            # 尝试加法
            count += Find_expression(target, end, total + num, expr + "+" + part)

            # 尝试减法
            count += Find_expression(target, end, total - num, expr + "-" + part)

    return count


# 统计目标值从1到100对应的表达式数量
Total_solutions = []

for i in range(1, 101):
    count = Find_expression(i)
    Total_solutions.append(count)

# 找到最大值和最小值
max_count = max(Total_solutions)
min_count = min(Total_solutions)

# 找到对应的目标值，考虑可能有多个最大值或最小值
max_numbers = [i + 1 for i, count in enumerate(Total_solutions) if count == max_count]
min_numbers = [i + 1 for i, count in enumerate(Total_solutions) if count == min_count]

print("最大解数：", max_count, "对应目标值：", max_numbers)
print("最小解数：", min_count, "对应目标值：", min_numbers)

# 绘制折线图
plt.figure(figsize=(12, 5))
plt.plot(range(1, 101), Total_solutions, marker='o', markersize=3)

plt.xlabel("Target Number")
plt.ylabel("Number of Solutions")
plt.title("Number of Solutions for Targets 1 to 100")
plt.grid(True, alpha=0.3)

plt.show()
```
