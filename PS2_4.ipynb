```python
# import pandas 
import pandas as pd
# import numpy
import numpy as np
# import matplotlib
from matplotlib import pyplot as plt
# make plots appear and be stored within the notebook
%matplotlib inline
```

```python
data = pd.read_csv('/Users/ruochenliu/Documents/ESE5023/Assignment_2/Q4_CentralPark_Monthly_Temperature_1998_2022.csv')
```

```python
#this is assignment 4.1
print(data[['DATE', 'TAVG']].isna().sum())

data = data.dropna(subset=['DATE', 'TAVG'])
data['DATE'] = pd.to_datetime(data['DATE'])

display(data.head())
```

```python
#this is assignment 4.2
data.plot(x='DATE', y='TAVG', figsize=(10, 4), xlabel='Time', ylabel='Temperature (°C)', title='Monthly Temperature in Central Park (1998–2022)')
```

```python
#this is assginment 4.3
mean = data['TAVG'].mean()
median = data['TAVG'].median()
minimum = data['TAVG'].min()
maximum = data['TAVG'].max()
std = data['TAVG'].std()

print('Mean:', mean)
print('Median:', median)
print('Minimum:', minimum)
print('Maximum:', maximum)
print('Standard deviation:', std)
```
