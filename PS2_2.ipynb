```python
# import pandas 
import pandas as pd
# import numpy
import numpy as np
# import matplotlib
from matplotlib import pyplot as plt
# make plots appear and be stored within the notebook
%matplotlib inline
```

```python
weather = pd.read_csv("/Users/ruochenliu/Documents/ESE5023/Assignment_2/Baoan_Weather_1998_2022.csv")
```

```python
weather['Temperature'] = weather['TMP'].str.split(',').str[0] #这两行是在拆分原始数据，首先用逗号拆分，str[0]是索引第一个数字就是温度
weather['Quality'] = weather['TMP'].str.split(',').str[1]#和上面一行同理

weather = weather[weather['Temperature'] != '+9999'] #根据自己选择的筛选条件，依据POS88-92，筛选掉缺失的数据

weather_new = weather[weather['Quality'].isin(['1', '5'])].copy() #索引quality那一列进行筛选，isin表示只筛选出质量码为1或者5的值

weather_new['Temperature'] = weather_new['Temperature'].astype(float) / 10 #索引出温度值那一列，根据POS88-92将温度先转换格式再换算为摄氏度

weather_new['DATE'] = pd.to_datetime(weather_new['DATE']) #这三列是索引时间这列然后分别提取出年和月为了便于计算平均温度
weather_new['Year'] = weather_new['DATE'].dt.year
weather_new['Month'] = weather_new['DATE'].dt.month

monthly_temperature = weather_new.groupby(['Year', 'Month'])['Temperature'].mean() #先用groupby函数分组，然后对温度求平均值

monthly_temperature.plot(figsize=(12, 4),xlabel='Year, Month',ylabel='Temperature (°C)',title='Monthly Average Air Temperature in Shenzhen (1998–2022)')
plt.show() # 最后画图
```
