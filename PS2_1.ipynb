```python
# import pandas 
import pandas as pd
# import numpy
import numpy as np
# import matplotlib
from matplotlib import pyplot as plt
# make plots appear and be stored within the notebook
%matplotlib inline
```

```python
Sig_Eqs = pd.read_csv(r"/Users/ruochenliu/Documents/ESE5023/Assignment_2/earthquakes-2026-10-14_21-39-07_+0800.tsv", sep="\t")
```

```python
#this is assignment 1.1
top_20_deaths = Sig_Eqs.groupby("Country")["Deaths"].sum().sort_values(ascending=False).head(20)
#‘groupby是将所有的数据按照国家分类，后面方括号里面的‘death’表示将其中死亡数这列数据提出来，然后对死亡数进行sum加和运算，最后再将这些算好的数据排序，默认是从小到大这里false表示从大到小，head函数表示选择前20个
print(top_20_deaths) #表示将刚刚分好类做好统计的数据打印出来
```

```python
# this is assignment 1.2
ms1 = Sig_Eqs[Sig_Eqs['Ms'] > 3.0] #这一行表示创建一个名为ms1的新的统计表，第一层方括号里面的内容表示选取‘Ms'那一列的数据，提取出大于3.0的数据，在筛选后再索引这些数据
annual_counts = ms1.groupby('Year').size()  # 这一行表示按年计数
print(annual_counts)#将按年计数的结果打印出来
```

```python
annual_counts.plot(xlabel='Year', ylabel='Number of earthquakes', title='Annual earthquake counts (Ms > 3.0)')
#将筛选好计好数的画图，x轴是年份，y轴是次数，标题为Ms>3.0的年度地震次数统计
```

```python
# this is assignment 1.3
def CountEq_LargestEq(country): #这里首先先定义函数
    data = Sig_Eqs[Sig_Eqs['Country'] == country] #从原始数据中索引出国家，提取出该国家的所有数据
    count = len(data) #统计该国家有记录的行数，相当于记录发生了多少次地震，也就是地震的次数

    max_ms = data['Ms'].max() #Ms表示地震强度，根据地震强度选出该国家发生过的最大的一次地震
    largest = data[data['Ms'] == max_ms] #筛选所有 Ms等于最大值的记录

    return count, largest[['Year', 'Mo', 'Dy', 'Location Name']]# 返回题目要求输出的两个结果，地震总次数，以及最大地震的年月日和地点

results = pd.DataFrame() #创建一个空表，用来存放所有国家的结果

for country in Sig_Eqs['Country'].value_counts().index: #这里用for循环遍历所有国家，index函数这里意思是取国家名，value_counts统计各国记录数，并从多到少排列
    count, largest = CountEq_LargestEq(country)# 调用函数，用 count 和 largest 分别接收函数返回的两个结果

    table = largest.copy()
    table['Country'] = country # 添加 Country 列，该列每一行都填写当前国家名
    table['Count'] = count  # 添加 Count 列，该列每一行都填写该国地震总次数

    results = pd.concat([results, table])  # 输出添加好的总结果

results = results[['Country', 'Count', 'Year', 'Mo', 'Dy', 'Location Name']] # 调整列的顺序
results[['Year', 'Mo', 'Dy']] = results[['Year', 'Mo', 'Dy']].astype('Int64') # 年、月、日显示为整数

pd.set_option('display.max_rows', None)
display(results.reset_index(drop=True))
```
